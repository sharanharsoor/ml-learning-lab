## 1. Read one row

`count` is the number of rentals in the hour. Weather and workday are category codes. For example, weather code 2 represents mist or clouds. Temperature is in degrees Celsius after multiplying the stored value by 41. Use only information available at prediction time to build a forecast.

```python
profile = pd.DataFrame(data["profile"])
profile.loc[profile["hour"] == 8, ["date", "hour", "workday", "weather", "temperature", "count"]]
```

## 2. Describe the five-hour profile

Start with the recorded counts. The mean divides their sum by five. The median is the third value in the sorted list. Variance averages squared distances from the mean. Standard deviation is its square root, which returns us to rentals.

Set `try_large_count = True` to replace the measured 779 with an illustrative 2,500 and run the cell again. Which measure of the center moves?

```python
try_large_count = False
counts = profile["count"].copy()
if try_large_count:
    counts.loc[profile["hour"] == 8] = 2500

mean = counts.mean()
median = counts.median()
distance = counts - mean
squared_distance = distance ** 2
variance = squared_distance.mean()
sd = variance ** 0.5

display(pd.DataFrame({
    "hour": profile["hour"],
    "count": counts,
    "count - mean": distance,
    "squared distance": squared_distance,
}))
print("Sorted counts:", list(counts.sort_values()))
print(f"Mean = {mean:,.2f} rentals")
print(f"Median = {median:,.2f} rentals")
print(f"Variance = {sum(squared_distance):,.2f} / {len(counts)} = {variance:,.2f} rentals²")
print(f"Standard deviation = {sd:,.2f} rentals")
```

## 3. Hold the hour fixed and look across many days

Each row describes a recorded 08:00 hour. The group summary compares workdays and non-workdays. Switch `day_type` to look at each group with the same chart axes. A plotted association gives a reason to investigate possible explanations.

```python
columns = ["date", "count", "temperature"]
workdays = pd.DataFrame(data["groups"]["workday"], columns=columns)
non_workdays = pd.DataFrame(data["groups"]["nonWorkday"], columns=columns)
eight_am = pd.concat([
    workdays.assign(day_type="Workday"),
    non_workdays.assign(day_type="Non-workday"),
], ignore_index=True)

display(eight_am.groupby("day_type")["count"].agg(["size", "mean", "median"]).round(2))
workday_forecast = round(workdays["count"].mean())
print(f"Workday forecast = round({workdays['count'].sum():,} / {len(workdays)}) = {workday_forecast}")
```

```python
day_type = "Workday"  # try "Non-workday"
recorded = eight_am.loc[eight_am["day_type"] == day_type]
edges = list(range(0, 901, 100))

try:
    import matplotlib.pyplot as plt
except ImportError:
    # Show the same histogram counts in a table if this local Python lacks Matplotlib.
    ranges = pd.cut(recorded["count"], bins=edges, right=False)
    display(ranges.value_counts(sort=False).rename("recorded hours"))
else:
    fig, axes = plt.subplots(1, 2, figsize=(11, 4))
    axes[0].hist(recorded["count"], bins=edges, color="#276e60", edgecolor="white")
    axes[0].set(xlim=(0, 900), xlabel="Rentals", ylabel="Recorded hours", title=day_type + " 08:00 histogram")
    axes[1].scatter(recorded["temperature"], recorded["count"], s=13, alpha=.42, color="#276e60")
    axes[1].set(xlim=(0, 35), ylim=(0, 900), xlabel="Recorded temperature (°C)",
                ylabel="Rentals", title=day_type + " 08:00 scatterplot")
    fig.tight_layout()
    plt.show()
```

## 4. Check what was recorded

A blank field can show up in a row that exists. To detect missing hours, build the expected date-hour keys separately and compare them with the keys in the file. An absent hour deserves a check in the data source before interpreting it as zero rentals.

The stored audit is reproducible from the original archive. Set `verify_from_source` to `True` to download it in Colab, or give `archive_path` the path to the ZIP you already have.

```python
verify_from_source = False
archive_path = ""  # optional path to an existing UCI Bike Sharing ZIP
audit = data["audit"]

if verify_from_source:
    import hashlib
    from io import BytesIO
    from urllib.request import urlopen
    from zipfile import ZipFile

    url = "https://archive.ics.uci.edu/static/public/275/bike+sharing+dataset.zip"
    if archive_path:
        archive_bytes = Path(archive_path).read_bytes()
    else:
        with urlopen(url, timeout=20) as response:
            archive_bytes = response.read()
    with ZipFile(BytesIO(archive_bytes)) as archive:
        raw = archive.read("hour.csv")
    assert hashlib.sha256(raw).hexdigest() == data["source_sha256"]
    hours = pd.read_csv(BytesIO(raw), keep_default_na=False)
    training = hours.loc[hours["dteday"] < "2012-11-01"]
    keys = set(zip(training["dteday"], training["hr"]))
    dates = pd.date_range(training["dteday"].min(), training["dteday"].max())
    expected = {(day.date().isoformat(), hour) for day in dates for hour in range(24)}
    measured = {
        "start": dates[0].date().isoformat(),
        "end": dates[-1].date().isoformat(),
        "rows": len(training),
        "blankCells": int(training.eq("").sum().sum()),
        "duplicateKeys": len(training) - len(keys),
        "expectedHours": len(expected),
        "absentHours": len(expected - keys),
    }
    assert audit == measured
    print("Verified source audit against hour.csv")

print(f"{audit['rows']:,} rows across {audit['expectedHours']:,} possible hours")
print(f"{audit['blankCells']} blank fields")
print(f"{audit['duplicateKeys']} extra rows sharing a date-hour key")
print(f"{audit['absentHours']} possible hours with no record")
```

Find the most common 100-rental range for each day type. Use that result to explain why calendar information is a sensible feature to try. When ready, run the cell to check your result against the two distributions.

```python
rental_range = pd.cut(eight_am["count"], bins=edges, right=False)
frequencies = eight_am.assign(rental_range=rental_range).groupby(
    ["day_type", "rental_range"], observed=True
).size().rename("recorded_hours").reset_index()

winners = frequencies.loc[
    frequencies["recorded_hours"].eq(frequencies.groupby("day_type")["recorded_hours"].transform("max"))
]
display(winners)
print("Day type is known before the forecast and the training groups have different 08:00 demand patterns.")
```
