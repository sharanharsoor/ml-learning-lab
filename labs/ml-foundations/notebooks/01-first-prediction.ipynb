# Your first ML prediction

It is just before the 08:00 hour on November 1, 2012. Estimate the coming hour’s system-wide rentals for Capital Bikeshare in Washington, DC.

We know the hour and the workday status in advance. These are the **features**. The **target** is the number of rentals we observe after the hour. Run the first cell to see a small set of historical rows.


```python
# Each tuple shows its date, hour, workday flag and observed rentals.
history = [
    ('2012-10-22', 8, 1, 728),
    ('2012-10-24', 8, 1, 801),
    ('2012-10-26', 8, 1, 734),
    ('2012-10-27', 8, 0, 213),
    ('2012-10-28', 8, 0, 173),
    ('2012-10-31', 8, 1, 621),
]
print("Date        Hour  Workday  Rentals")
for date, hour, workingday, rentals in history:
    print(f"{date}     {hour}        {workingday}      {rentals}")
```

## Fit a mean, then use it to predict

Each source row describes one hour. Workday is 1 on weekdays that are not holidays and 0 otherwise.

Use all training records through October 31, 2012 that match 08:00 and workday 1. We show six rows above to explain the inputs. The historical group actually contains 456 hours and recorded 214,195 rentals. Divide the recorded total by the number of hours and round to the nearest rental.


```python
total_rentals = 214195
matching_hours = 456
mean_rentals = total_rentals / matching_hours
forecast = round(mean_rentals)

print(f"{total_rentals:,} ÷ {matching_hours} = {mean_rentals:.2f} rentals per hour")
print(f"Forecast: {forecast} rentals")
assert forecast == 470
```

## Evaluate after the hour ends

The recorded count is now available. Compute a **residual** by subtracting the forecast from the recorded count. First predict its sign. A positive residual means the recorded count was higher than the forecast.


```python
actual_rentals = 680
residual = actual_rentals - forecast
print(f"{actual_rentals} − {forecast} = {residual:+} rentals")
assert residual == 210
```

## Try 09:00

Use the same rule on the next hour. Past 09:00 workday hours recorded 108,510 rentals across 456 hours. The held-out 09:00 hour recorded 331 rentals. Make a rounded forecast and find the residual. Did the forecast run high or low?
